# 键盘控制面板
先启动底盘服务；可使用 `--simulate --enable-motion`。键盘由 **kernel 所在桌面的 pynput** 捕获，浏览器仅显示状态。这需要可访问且能接收实际键盘事件的桌面会话；普通 SSH 请运行 `python -m rsim.apps.keyboard_control --input terminal`，X 转发本身并不保证能捕获本地终端按键。

复制 `keyboard.example.yaml` 到项目 `configs/keyboard.yaml` 后调整模拟参数。默认 `DRY_RUN=True`，模型响应按键但只发送零速度。W/S 加减线速度，A/D 加减角速度，松键渐停，空格立即制动，Esc 退出。面板 Brake 按钮保持制动，按下再松开空格可解除。

```python
import os
from pathlib import Path
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'pyproject.toml').is_file() and (p / 'rsim').is_dir())
os.environ['NOTEBOOK_ASSETS_ROOT'] = str(ROOT / 'assets')
from rsim.apps.notebook_control import KeyboardDashboard
from rsim.transport.descriptor import TransportConfig
from IPython.display import display
DRY_RUN = False # True
panel = KeyboardDashboard(ROOT / 'configs' / 'keyboard.yaml',
    name=os.environ.get('RSIM_CHASSIS_NAME', 'chassis'),
    transport=TransportConfig(domain_id=int(os.environ.get('RSIM_CONTROL_DOMAIN', '0'))),
    dry_run=DRY_RUN)
display(await panel.start())
```

启动单元格立即返回，采集、控制和面板刷新作为 task 留在 Notebook 已有事件循环中运行。完成后执行下面的清理单元格，或点击 Stop and close。

```python
await panel.close()
```
